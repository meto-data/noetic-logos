```python
import numpy as np
import pandas as pd
```

```python

# Rastgele veri seti oluşturma (Gürültü eklenmiş)
np.random.seed(3)
data1 = {
    "ID": range(1, 51),
    "Maas": np.random.randint(100, 1000, size=50) + np.random.normal(0, 50),  
}

data = { "ID": range(1, 41), # 1'den 50'e kadar ID sütunu 
         "Kategori": np.random.choice(["A", "B", "C", "D", "E"], size=40),
}


df1 = pd.DataFrame(data)
df2 = pd.DataFrame(data1)

print(df1.head(5))
print(df2.head(5))

```

```python
# Dikey birleştirme   --> Satır Bazlı - axis=0
df_concat = pd.concat([df1, df2], axis=0, ignore_index=True)

print(df_concat)
```

```python
# Yatay birleştirme   --> Sütun Bazlı - axis=1
df_concat = pd.concat([df1, df2], axis=0, ignore_index=True)

print(df_concat)
```

```python
# Örnek iki veri seti
d1 = pd.DataFrame({'ID': [1, 2, 3], 'Ad': ['Ali', 'Ayşe', 'Mehmet'], 'Yaş': [25, 30, 22]})
d2 = pd.DataFrame({'ID': [4, 5, 6], 'Ad': ['Fatma', 'Hasan', 'Zeynep'], 'Yaş': [27, 24, 29]})

# Dikey birleştirme
d_concat = pd.concat([d1, d2], axis=0, ignore_index=True)   #ignore_index=True kullanılarak indeks sıfırdan yeniden numaralandırıldı.

print(d_concat)
```

```python

# Ortak sütuna göre birleştirme
df_merge = pd.merge(df1, df2, on="ID")

print(df_merge)
```
