```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x = [1,2,3,4]
y = [1,4,9,16]


plt.plot(x,y)
plt.show()
```

```python
plt.title("İlk grafiğimiz!")
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.plot(x,y)
plt.show()
```

```python

# Normalde çentikler otomatik olarak yerleştirilir
# Kendiniz de çentiklerin yerini tanımlayabilirsiniz

plt.xticks([1,2,3,4])   #küsüratları da gösterebilir!
plt.yticks([1,4,9,16])
plt.title("İlk grafiğimiz!")
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.plot(x,y)
plt.show()
```

<H3> Lejant koyma  </H3>


```python
plt.plot(x,y, label='x^2', color='green')
plt.xticks([1,2,3,4,5])   
plt.yticks([1,4,9,16,25])
plt.title("İlk grafiğimiz!")
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.show()
```

```python
# linewidth ekleme
plt.plot(x,y, label='x^2', color='blue', linewidth = 2, linestyle = '--', marker='.')
plt.xticks([1,2,3,4,5])   
plt.yticks([1,4,9,16,25])
plt.title("İlk grafiğimiz!")
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.show()
```

<H3>Barchart</H3>

```python
x = ['Ankara', 'İstanbul', 'İzmir', 'Aydın', 'Sakarya', 'Edirne', 'Düzce']
y = [120, 178, 87, 23, 17, 35, 29]

plt.bar(x,y, color='pink')
plt.xlabel('Kategori')
plt.ylabel('Değerler')
plt.show()
```

```python

# Eğer çubuklardan bir tanesini diğerlerinden farklılaştırmak için içini boyamak isterseniz:
x = ['Ankara', 'İstanbul', 'İzmir', 'Kastamonu', 'Sakarya', 'Edirne', 'Düzce']
y = [120, 178, 87, 23, 17, 35, 29]

cubuklar = plt.bar(x,y, color='pink')
cubuklar[1].set_hatch('+')
cubuklar[0].set_hatch('.')
cubuklar[4].set_hatch('/')
plt.show()




```

```python
import matplotlib.pyplot as plt

dilimler = [20, 30, 25, 25]
etiketler = ['Grup A', 'Grup B', 'Grup C', 'Grup D']

plt.pie(dilimler, labels=etiketler, autopct='%1.1f%%', startangle=140)
plt.title('Pasta Grafiği Örneği')
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

veri = np.random.randn(1000)

plt.hist(veri, bins=30, color='purple', edgecolor='black')
plt.title('Histogram Örneği')
plt.xlabel('Değer Aralığı')
plt.ylabel('Frekans')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color='green')
plt.title('Scatter Grafiği Örneği')
plt.xlabel('X ekseni')
plt.ylabel('Y ekseni')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 5, 1)
y1 = [1, 2, 4, 3, 5]
y2 = [2, 3, 5, 4, 6]

plt.fill_between(x, y1, color="skyblue", alpha=0.4)
plt.fill_between(x, y2, color="lightgreen", alpha=0.4)
plt.plot(x, y1, label='Y1', color='blue')
plt.plot(x, y2, label='Y2', color='green')

plt.title('Alan Grafiği Örneği')
plt.xlabel('X ekseni')
plt.ylabel('Y ekseni')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, (ax1, ax2) = plt.subplots(2, 1)

ax1.plot(x, y1, color='red')
ax1.set_title('Sinüs Grafiği')

ax2.plot(x, y2, color='blue')
ax2.set_title('Kosinüs Grafiği')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

degerler = [5, 3, 4, 2, 4, 5]
etiketler = np.array(['A', 'B', 'C', 'D', 'E', 'F'])

angles = np.linspace(0, 2 * np.pi, len(degerler), endpoint=False).tolist()
degerler += degerler[:1]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(6,6), subplot_kw=dict(polar=True))
ax.fill(angles, degerler, color='blue', alpha=0.25)
ax.plot(angles, degerler, color='blue')

ax.set_xticks(angles[:-1])
ax.set_xticklabels(etiketler)

plt.title('Radar Grafiği Örneği')
plt.show()
```
