```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

df = pd.DataFrame({"Gelir": [2500, 3000, 4000, 10000, 15000]})
scaler = MinMaxScaler()
df["Gelir_Normalize"] = scaler.fit_transform(df[["Gelir"]])

print(df)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df["Gelir_Standardize"] = scaler.fit_transform(df[["Gelir"]])

print(df)
```

```python
import numpy as np

df["Gelir_Log"] = np.log(df["Gelir"])
print(df)
```

```python
df_kategori = pd.DataFrame({"Şehir": ["İstanbul", "Ankara", "İzmir", "İstanbul"]})
df_encoded = pd.get_dummies(df_kategori, columns=["Şehir"])
print(df_encoded)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
df_kategori["Şehir_Encoded"] = encoder.fit_transform(df_kategori["Şehir"])
print(df_kategori)
```
